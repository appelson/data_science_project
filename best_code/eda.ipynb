```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_clean = pd.read_csv('/Users/leachen/Docs/uni/masters/stanford_mse/fall25/mse226/data_science_project/cleaned_data/cleaned_data.csv')
df_clean.head()
```

```python
## Correlations numerical
corr_matrix = df_clean.corr(numeric_only=True)
plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap='coolwarm')
plt.title('Correlation Matrix of Numerical Features')
plt.show()
```

```python
## Distributional plots
def plot_distribution(df, column_name):
    plt.figure(figsize=(8, 6))
    sns.histplot(df[column_name], kde=True, bins=30)
    plt.title(f'Distribution of {column_name}')
    plt.xlabel(column_name)
    plt.ylabel('Frequency')
    plt.show()

plot_distribution(df_clean[df_clean['locale_code']=='Town'], 'print_volumes')
```
